# სასტუმროს AI ხმოვანი ასისტენტი — სრული გიდი LangGraph-ზე

## მოგესალმებით! ამ სახელმძღვანელოში ჩვენ ნულიდან ავაგებთ სისტემას, რომელიც სტუმარს ხმით ესაუბრება ქართულ ენაზე, იღებს შეკვეთებს, ამოწმებს სასტუმროს მონაცემთა ბაზას (PMS) და შეცდომის შემთხვევაში იღებს გონივრულ გადაწყვეტილებებს.

## 1. რა არის State (მდგომარეობა)?

წარმოიდგინეთ ოფიციანტი, რომელიც თქვენს მაგიდასთან მოდის ბლოკნოტით.

სანამ თქვენ ელაპარაკებით, ის ბლოკნოტში იწერს: თქვენს სახელს, ოთახის ნომერს, რა შეუკვეთეთ და რა შეეშალა სამზარეულოს.

State არის სწორედ ეს ბლოკნოტი.

ჩვეულებრივი LLM (მაგ. ChatGPT) ყოველი მოთხოვნის შემდეგ ივიწყებს ყველაფერს (Stateless). LangGraph-ში State არის ერთი საერთო მეხსიერების ლექსიკონი (Dictionary), რომელსაც ყველა ნაბიჯი კითხულობს და ანახლებს.
